```python
import requests
```

```python
 url = 'http://openapi.seoul.go.kr:8088/4875635778616c733132374b73524659/json/SebcCollegeInfoKor/1/100/'
```

```python
response = requests.get(url)
```

```python
data = response.json()
data
```

```python
import pandas as pd
temp = []
for d in data['SebcCollegeInfoKor']['row']:
    temp.append(d)
unl = pd.DataFrame(temp)
unl.head(3)
```

```python
l_url = 'http://openapi.seoul.go.kr:8088/4875635778616c733132374b73524659/json/SeoulPublicLibraryInfo/1/100/'
```

```python
l_response = requests.get(l_url)
```

```python
l_data = l_response.json()
l_data
```

```python
import pandas as pd
temp = []
for d in l_data['SeoulPublicLibraryInfo']['row']:
    temp.append(d)
```

```python
l_unl = pd.DataFrame(temp)
l_unl.head(3)
```

```python
import folium
anMap = folium.Map(location=[37.4733061524019,127.099727862691],
                   zoom_start=11,
                   zoom_control=False,
                   control_scale=True)
anMap
```

```python
from folium.plugins import MarkerCluster
mc = MarkerCluster().add_to(anMap)

for n, row in l_unl.iterrows():
  folium.Marker(location = [row['XCNTS'],row['YDNTS']], popup = folium.Popup(row['LBRRY_NAME'],max_width=300)).add_to(mc)
anMap
```
